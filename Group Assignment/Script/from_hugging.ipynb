```python
from datasets import load_dataset

ds = load_dataset("Nicolybgs/healthcare_data")
```

```python
import os

data_dir = os.path.join("..", "Data")
os.makedirs(data_dir, exist_ok=True)

for split_name, split_ds in ds.items():
    out_path = os.path.join(data_dir, f"{split_name}.parquet")
    split_ds.to_parquet(out_path)
    print(f"Saved {split_name} ({split_ds.num_rows} rows) -> {out_path}")
```

```python
import pandas as pd

df = pd.read_parquet(os.path.join(data_dir, "train.parquet"))
df.head(20)
```

```python
num_duplicate_ids = df["patientid"].duplicated().sum()
print(f"Total rows: {len(df)}")
print(f"Unique patientid: {df['patientid'].nunique()}")
print(f"Duplicated patientid rows: {num_duplicate_ids}")

duplicate_patient_rows = df[df["patientid"].duplicated(keep=False)].sort_values("patientid")
duplicate_patient_rows.head(20)
```
