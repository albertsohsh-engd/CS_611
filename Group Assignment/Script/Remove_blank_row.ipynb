```python
import os
import pandas as pd

data_dir = os.path.join("..", "Data")
processed_dir = os.path.join("..", "Processed Data")
os.makedirs(processed_dir, exist_ok=True)
```

```python
order_info = pd.read_csv(os.path.join(data_dir, "Order Info.csv"))
print("Before:", order_info.shape)

order_info_clean = order_info.dropna(how="all")
print("After:", order_info_clean.shape)

order_info_clean.to_csv(os.path.join(processed_dir, "Order Info_remove_blank.csv"), index=False)
```

```python
locker_order_info = pd.read_csv(os.path.join(data_dir, "locker_order_info.csv"))
print("Before:", locker_order_info.shape)

locker_order_info_clean = locker_order_info.dropna(how="all")
print("After:", locker_order_info_clean.shape)

locker_order_info_clean.to_csv(os.path.join(processed_dir, "locker_order_info_remove_blank.csv"), index=False)
```
